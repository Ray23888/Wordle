```python
from wordle_result import LetterAnalysis

def analyze(word, target):
    '''
    With the parameters of a guessed word and the solution word, the function checks each letter of the guess against the solution and adds the index of the guess to a dictionary with a value of how it matches (exact, partial, or none).

    Parameters:
    word (str): The guessed word.
    target (str): The solution word.

    Returns:
    dict: A dictionary of indexes of the guess word with how they match up to the solution word.
    '''
    # Define constants for different match types
    match = LetterAnalysis.EXACT_MATCH
    partial = LetterAnalysis.PARTIAL_MATCH
    no = LetterAnalysis.NO_MATCH

    # Initialize an empty dictionary to store the match results for each index in the guessed word
    guess = {}

    # Iterate through each index and character in the guessed word
    for i, v in enumerate(word):
        # Check if the character at the current index in the guessed word matches the character at the same index in the solution word
        if v == target[i]:
            guess[i] = match  # If yes, it's an exact match
        # Check if the character at the current index in the guessed word exists anywhere in the solution word
        elif v in target:
            guess[i] = partial  # If yes, it's a partial match
        else:
            guess[i] = no  # If neither, it's no match

        # Remove the first occurrence of the guessed character in the remaining portion of the solution word
        target = target.replace(v, " ", 1)

    # Return the dictionary containing match results for each index in the guessed word
    return guess
```

```python
import random
from stats import Statistics
from core import analyze
from coloring import colored_word

# Create an instance of the Statistics class
statistics = Statistics()

def run_game(file_name):
    '''Define a function called run_game that takes in the dictionary file.'''
    # Open the file and create a list of words from each line
    with open(file_name, "r") as file:
        dictionary = [word.strip() for word in file.readlines()]

    # Flag to control whether the player wants to play again
    game_status = "y"

    # Main game loop
    while game_status == "y":
        # Choose a random word from the dictionary as the secret word
        secret_word = random.choice(dictionary)

        # Allow the user to guess up to 6 times
        guesses = 6
        while guesses > 0:
            # Ask the user for input and capitalize each letter
            user_input = input("Enter a Guess: ").upper()

            # Validate user input
            if len(user_input) != 5 or user_input not in dictionary:
                print("invalid")
            elif user_input == secret_word:
                # If the user guessed correctly, update statistics and break the loop
                statistics.add_won_game(6 - guesses + 1)
                analyzed_word = analyze(user_input, secret_word)
                print(colored_word(user_input, analyzed_word))
                print("Congratulations! You have won.")
                break
            else:
                # If the guess is incorrect, update statistics and continue
                analyzed_word = analyze(user_input, secret_word)
                print(colored_word(user_input, analyzed_word))
                guesses -= 1

        if guesses == 0:
            # If the user runs out of guesses, update statistics
            statistics.add_lost_game()
            print("You lost.")

        # Get and print statistics
        statistics.get_guess_distribution()
        statistics.get_success_metrics()
        statistics.print_stats()

        # Ask the user if they want to play again
        game_status = input("Enter 'y' to play again or 'n' to quit ").lower()

        # Exit the loop if the user does not want to play again
        if game_status == "n":
            break
```
